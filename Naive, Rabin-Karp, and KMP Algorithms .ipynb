```python
import random
import string

# -----------------------------
# Generate Random Text
# -----------------------------
def generate_text(length):
    return ''.join(random.choice(string.ascii_lowercase) for _ in range(length))

# -----------------------------
# Naive String Matching
# -----------------------------
def naive_search(text, pattern):
    n = len(text)
    m = len(pattern)
    comparisons = 0
    occurrences = []

    for i in range(n - m + 1):
        j = 0
        while j < m:
            comparisons += 1
            if text[i + j] != pattern[j]:
                break
            j += 1
        if j == m:
            occurrences.append(i)

    return occurrences, comparisons


# -----------------------------
# KMP Algorithm
# -----------------------------
def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1

    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        elif length != 0:
            length = lps[length - 1]
        else:
            lps[i] = 0
            i += 1

    return lps


def kmp_search(text, pattern):
    n = len(text)
    m = len(pattern)

    lps = compute_lps(pattern)

    comparisons = 0
    occurrences = []

    i = 0
    j = 0

    while i < n:
        comparisons += 1

        if text[i] == pattern[j]:
            i += 1
            j += 1

            if j == m:
                occurrences.append(i - j)
                j = lps[j - 1]

        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    return occurrences, comparisons


# -----------------------------
# Rabin-Karp Algorithm
# -----------------------------
def rabin_karp(text, pattern, prime=101):
    n = len(text)
    m = len(pattern)

    d = 256

    h = pow(d, m - 1, prime)

    p_hash = 0
    t_hash = 0

    comparisons = 0
    occurrences = []

    for i in range(m):
        p_hash = (d * p_hash + ord(pattern[i])) % prime
        t_hash = (d * t_hash + ord(text[i])) % prime

    for i in range(n - m + 1):

        if p_hash == t_hash:
            match = True

            for j in range(m):
                comparisons += 1
                if text[i + j] != pattern[j]:
                    match = False
                    break

            if match:
                occurrences.append(i)

        if i < n - m:
            t_hash = (d * (t_hash - ord(text[i]) * h) + ord(text[i + m])) % prime

            if t_hash < 0:
                t_hash += prime

    return occurrences, comparisons


# -----------------------------
# Main Program
# -----------------------------
TEXT_LENGTH = 10000

text = generate_text(TEXT_LENGTH)

pattern_lengths = [76, 15, 34, 60]

print("Text Length =", len(text))
print()

print("{:<12}{:<20}{:<20}{:<20}".format(
    "Pattern", "Naive", "KMP", "Rabin-Karp"))
print("-" * 75)

for length in pattern_lengths:

    # Choose a pattern that definitely exists in the text
    start = random.randint(0, TEXT_LENGTH - length)
    pattern = text[start:start + length]

    _, naive_cmp = naive_search(text, pattern)
    _, kmp_cmp = kmp_search(text, pattern)
    _, rk_cmp = rabin_karp(text, pattern)

    print("{:<12}{:<20}{:<20}{:<20}".format(
        length,
        naive_cmp,
        kmp_cmp,
        rk_cmp))
```
